```python
import pandas as pd
import re
from pathlib import Path
import csv

def normalize_name(name: str) -> str:
    # Remove BOM/whitespace, unify spaces/underscores, drop units in brackets into suffix-less form
    name = name.strip()
    # Remove bracketed units, keep a hint if helpful
    # e.g., "RecordingTime [ms]" -> "RecordingTime"
    name = re.sub(r"\[[^\]]*\]", "", name)
    # Collapse whitespace and punctuation to underscores
    name = re.sub(r"[^\w]+", "_", name)
    name = name.strip("_").lower()
    return name

# Map verbose original labels to a compact canonical schema
CANONICAL_MAP = {
    # basics
    "recordingtime": "recording_time_ms",
    "recordingtime_ms": "recording_time_ms",
    "time_of_day_h_m_s_ms": "time_of_day",
    "trial": "trial",
    "stimulus": "stimulus",
    "export_start_trial_time": "export_start_ms",
    "export_start_trial_time_ms": "export_start_ms",
    "export_end_trial_time": "export_end_ms",
    "export_end_trial_time_ms": "export_end_ms",
    "participant": "participant",
    "tracking_ratio": "tracking_ratio_pct",
    "tracking_ratio_pct": "tracking_ratio_pct",
    "color": "color",

    # pupil diameters
    "pupil_diameter_right_mm": "pupil_r_mm",
    "pupil_diameter_left_mm": "pupil_l_mm",

    # point of regard (pixels)
    "point_of_regard_right_x_px": "por_r_x_px",
    "point_of_regard_right_y_px": "por_r_y_px",
    "point_of_regard_left_x_px": "por_l_x_px",
    "point_of_regard_left_y_px": "por_l_y_px",
}

# Obvious noise to drop (you can adjust this list)
DROP_PATTERNS = [
    r"^unnamed",                # Unnamed: 0, etc.
    r"^mouse_",                 # Mouse Position
    r"^scroll_",                # Scroll
    r"^port_status$",           # Port
    r"^content$",               # Content (verbatim filename already in stimulus)
    r"^gaze_vector_",           # 3D gaze vectors (if unused)
    r"^eye_position_",          # Eye position in mm (if unused)
    r"^pupil_position_",        # Pupil position in px (if unused)
    r"^aoi_",                   # AOI metadata (if not needed now)
    r"^annotation_",            # Annotation Name/Description/Tags
    r"^binocular$",             # stray label
    r"^groupe_d_enfants$",      # localized extra
    r"^category", r"^index_",   # right/left category/index columns
]

def should_drop(col: str) -> bool:
    return any(re.match(pat, col) for pat in DROP_PATTERNS)

def load_and_clean_csv(path) -> pd.DataFrame:
    # Let pandas sniff comma vs semicolon; handle BOM
    df = pd.read_csv(path, sep=None, engine="python", encoding="utf-8-sig", quoting=csv.QUOTE_MINIMAL)

    # Normalize headers
    norm_cols = [normalize_name(c) for c in df.columns]
    df.columns = norm_cols

    # Drop obvious junk (Unnamed, mouse, etc.)
    keep_cols = [c for c in df.columns if not should_drop(c)]
    df = df[keep_cols]

    # Canonical rename (best-effort)
    rename_map = {c: CANONICAL_MAP.get(c, c) for c in df.columns}
    df = df.rename(columns=rename_map)

    # Add source for traceability
    df["source_file"] = Path(path).name

    # Parse/convert key types
    if "recording_time_ms" in df.columns:
        # Sometimes these are floats; keep numeric
        df["recording_time_ms"] = pd.to_numeric(df["recording_time_ms"], errors="coerce")

    if "export_start_ms" in df.columns:
        df["export_start_ms"] = pd.to_numeric(df["export_start_ms"], errors="coerce")
    if "export_end_ms" in df.columns:
        df["export_end_ms"] = pd.to_numeric(df["export_end_ms"], errors="coerce")

    for col in ["pupil_r_mm", "pupil_l_mm", "por_r_x_px", "por_r_y_px", "por_l_x_px", "por_l_y_px", "tracking_ratio_pct"]:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors="coerce")

    # Participant as string ID
    if "participant" in df.columns:
        df["participant"] = df["participant"].astype(str).str.strip()

    # Tidy: remove fully empty columns that may remain after coercion
    df = df.dropna(axis=1, how="all")

    return df

# ---- Use on one file
# df = load_and_clean_csv("data/raw/your_file.csv")
# df.head()

# ---- Or: batch all files into a master
def build_master(folder="data/raw"):
    paths = sorted(Path(folder).glob("*.csv"))
    dfs = [load_and_clean_csv(p) for p in paths]
    master = pd.concat(dfs, ignore_index=True, sort=False)

    # Optional: keep only the canonical subset for analysis
    canonical_subset = [
        "source_file",
        "participant", "trial", "stimulus", "color",
        "recording_time_ms", "time_of_day",
        "export_start_ms", "export_end_ms",
        "tracking_ratio_pct",
        "pupil_r_mm", "pupil_l_mm",
        "por_r_x_px", "por_r_y_px", "por_l_x_px", "por_l_y_px",
    ]
    have = [c for c in canonical_subset if c in master.columns]
    master = master[have]

    return master

# Example:
# master = build_master("data/raw")
# master.to_parquet("data/processed/master.parquet", index=False)
# master.to_csv("data/processed/master.csv", index=False)
```

```python
# prefer using the notebook's load_and_clean_csv (defined in an earlier cell)
lookFor = "2"
matches = []

for i in range(1, 26):
    path = f"../data/{i}.csv"

    # try to use the canonical loader if available, fall back to raw read
    try:
        df_i = load_and_clean_csv(path)
    except NameError:
        # load_and_clean_csv not defined in session; do a tolerant read
        df_i = pd.read_csv(path, sep=None, engine="python", encoding="utf-8-sig")
    except Exception as e:
        print(f"Failed to read {path}: {e}")
        continue

    # find participant column robustly (normalized loader will provide 'participant')
    if "participant" in df_i.columns:
        part_col = "participant"
    else:
        # try case-insensitive match to common original header
        cols_lower = {c.lower(): c for c in df_i.columns}
        part_col = cols_lower.get("participant")



    subset = df_i[df_i[part_col] == lookFor]
    if not subset.empty:
        print(f"File {i} ({path}) has {len(subset)} rows with participant == {lookFor}")
        matches.append(subset)

if matches:
    all_matches = pd.concat(matches, ignore_index=True, sort=False)
    print("Total rows with participant == 'Unidentified(Neg)':", len(all_matches))
else:
    print("No matches found for", lookFor)
```

```python
print("Matches details:")
for i, match in enumerate(matches, start=1):
    print(f"Match {i}:")
    print(match.head(1).to_string()) 
```
